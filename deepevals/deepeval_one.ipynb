```python
import os
from langchain_groq import ChatGroq
from deepeval.models.base_model import DeepEvalBaseLLM


class Groqchatllm(DeepEvalBaseLLM):
    def __init__(self, model_name: str, api_key: str = None):
        self.model_name = model_name
        self.api_key = api_key_1
        self.model = ChatGroq(
            model=model_name,
            api_key=self.api_key,
            temperature=0.7,
            max_tokens=700,
        )

    def load_model(self):
        return self.model

    def generate(self, prompt: str) -> str:
        res = self.model.invoke(prompt)
        return res.content if hasattr(res, "content") else str(res)

    async def a_generate(self, prompt: str) -> str:
        res = await self.model.ainvoke(prompt)
        return res.content if hasattr(res, "content") else str(res)

    def get_model_name(self):
        return f"Groq-{self.model_name}"


# Usage
custom_model = Groqchatllm("llama-3.3-70b-versatile")
```

```python
from deepeval.test_case import LLMTestCase
from deepeval.metrics import AnswerRelevancyMetric

ans_relevancy_metrics = AnswerRelevancyMetric(model=custom_model)

test_case = LLMTestCase(
    input="who is india's highest run scorer in ODI?",
    actual_output= "Virat Kohli",
    expected_output="Rohit Sharma has highest ODI Score of 264",
    retrieval_context= ["Virat Kholi has scored 183 as his individual highest in One Day International."]
    
)


ans_relevancy_metrics.measure(test_case=test_case)

```

```python
ans_relevancy_metrics.score, ans_relevancy_metrics.reason
```

```python
from deepeval.metrics import GEval
from deepeval.test_case import SingleTurnParams

correctness_metric = GEval(
    name="Correctness",
    criteria="Determine whether the actual output is factually correct based on the expected output.",
    # NOTE: you can only provide either criteria or evaluation_steps, and not both
    evaluation_steps=[
        "Check whether the facts in 'actual output' contradicts any facts in 'expected output'",
        "You should also heavily penalize omission of detail",
        "Vague language, or contradicting OPINIONS, are OK",
    ],
    model=custom_model,
    evaluation_params=[
        SingleTurnParams.INPUT,
        SingleTurnParams.ACTUAL_OUTPUT,
        SingleTurnParams.EXPECTED_OUTPUT,
    ],
)
```

```python
correctness_metric.measure(test_case=test_case)
```

```python
from deepeval.metrics import BaseMetric
from deepeval.test_case import LLMTestCase


class CustomHybridMetrics(BaseMetric):
    def __init__(
        self,
        threshold: float = 0.5,
        verbose_logs: bool = True,
        strict_mode: bool = True,
    ):
        self.threshold = threshold
        self.verbose_logs = verbose_logs
        self.strict_mode = strict_mode
        self.error = None

    def measure(self, test_case: LLMTestCase) -> dict:
        try:
            return self._is_checked(test_case.actual_output, test_case.expected_output)
        except Exception as e:
            self.error = str(e)
            raise

    def _is_checked(self, x: str, y: str) -> float:
        if x == y:
            return {"score" : 1.0}
        return {"score": 0.0}
    
    
    
t1 = LLMTestCase(input="test_case", actual_output="1231", expected_output="1232")
cust = CustomHybridMetrics()

cust.measure(t1)    
    
```
